# Distribution Analysis

```python
SAMPLE_FRAC = 0.10
RANDOM_STATE = 42

analysis2018_sample = analysis2018.sample(
    frac=SAMPLE_FRAC,
    random_state=RANDOM_STATE
).copy()

analysis2019_sample = analysis2019.sample(
    frac=SAMPLE_FRAC,
    random_state=RANDOM_STATE
).copy()

analysis2018_sample["year"] = "2018"
analysis2019_sample["year"] = "2019"

analytic_sample = pd.concat(
    [analysis2018_sample, analysis2019_sample],
    axis=0
).copy()

print("2018 sample shape:", analysis2018_sample.shape)
print("2019 sample shape:", analysis2019_sample.shape)
print("Overall analytic sample shape:", analytic_sample.shape)
```

```python
AGE_COL = "Member_DOB_Age_Qty"
FEMALE_COL = "Gender_Categorical"
HTN_COL = "Member_Has_HTN_Diag_Ind"

PCP_COL = "Office_PCP_Count"
ER_COL = "ER_Count"

analytic_sample["Gender_Categorical"] = analytic_sample["Member_Gender_Current"].map({
    "Male": 0,
    "Female": 1
})
analysis2018_sample["Gender_Categorical"] = analysis2018_sample["Member_Gender_Current"].map({
    "Male": 0,
    "Female": 1
})
analysis2019_sample["Gender_Categorical"] = analysis2019_sample["Member_Gender_Current"].map({
    "Male": 0,
    "Female": 1
})
```

```python
print("Female values:")
print(analytic_sample[FEMALE_COL].value_counts(dropna=False).head())

print("\nHypertension values:")
print(analytic_sample[HTN_COL].value_counts(dropna=False).head())

print("\nAge summary:")
print(analytic_sample[AGE_COL].describe())

print("\nPCP summary:")
print(analytic_sample[PCP_COL].describe())

print("\nER summary:")
print(analytic_sample[ER_COL].describe())
```

```python
print("Female values:")
print(analysis2018_sample[FEMALE_COL].value_counts(dropna=False).head())

print("\nHypertension values:")
print(analysis2018_sample[HTN_COL].value_counts(dropna=False).head())

print("\nAge summary:")
print(analysis2018_sample[AGE_COL].describe())

print("\nPCP summary:")
print(analysis2018_sample[PCP_COL].describe())

print("\nER summary:")
print(analysis2018_sample[ER_COL].describe())
```

```python
print("Female values:")
print(analysis2019_sample[FEMALE_COL].value_counts(dropna=False).head())

print("\nHypertension values:")
print(analysis2019_sample[HTN_COL].value_counts(dropna=False).head())

print("\nAge summary:")
print(analysis2019_sample[AGE_COL].describe())

print("\nPCP summary:")
print(analysis2019_sample[PCP_COL].describe())

print("\nER summary:")
print(analysis2019_sample[ER_COL].describe())
```

```python
ID_COL = "Member_Wid"
n_obs = len(analytic_sample)

mean_age = analytic_sample[AGE_COL].mean()
sd_age = analytic_sample[AGE_COL].std()

female_n = int(analytic_sample[FEMALE_COL].sum())
female_pct = analytic_sample[FEMALE_COL].mean() * 100

htn_n = int(analytic_sample[HTN_COL].sum())
htn_pct = analytic_sample[HTN_COL].mean() * 100

# mean_pcp = analytic_sample[PCP_COL].mean()
# sd_pcp = analytic_sample[PCP_COL].std()

# mean_er = analytic_sample[ER_COL].mean()
# sd_er = analytic_sample[ER_COL].std()
annual_utilization = (
    analytic_sample
    .groupby([ID_COL, "year"], as_index=False)
    .agg(
        annual_pcp_count=(PCP_COL, "sum"),
        annual_er_count=(ER_COL, "sum")
    )
)

mean_annual_pcp = annual_utilization["annual_pcp_count"].mean()
sd_annual_pcp = annual_utilization["annual_pcp_count"].std()

mean_annual_er = annual_utilization["annual_er_count"].mean()
sd_annual_er = annual_utilization["annual_er_count"].std()

print(f"N observations: {n_obs:,}")
print(f"Mean age: {mean_age:.1f} ({sd_age:.1f})")
print(f"Female: {female_n:,} ({female_pct:.1f}%)")
print(f"Hypertension: {htn_n:,} ({htn_pct:.1f}%)")
print(f"Mean PCP count: {mean_annual_pcp:.3f} ({sd_annual_pcp:.3f})")
print(f"Mean ER count: {mean_annual_er:.3f} ({sd_annual_er:.3f})")
```
