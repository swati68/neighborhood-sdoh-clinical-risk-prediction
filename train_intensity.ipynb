```python
x_train_nb_sample = pd.read_csv("../data_processed/x_train_nb_sample.csv")
y_train_nb_sample = pd.read_csv("../data_processed/y_train_nb_sample.csv")

x_test_nb_sample = pd.read_csv("../data_processed/x_test_nb_sample.csv")
y_test_nb_sample = pd.read_csv("../data_processed/y_test_nb_sample.csv")
```

```python
x_train_nb = x_train_nb_sample[individual].astype(float)
x_test_nb = x_test_nb_sample[individual].astype(float)

y_train_pcp_count = y_train_nb_sample["pcp_intensity"].astype(int)
y_test_pcp_count = y_test_nb_sample["pcp_intensity"].astype(int)

# Add intercept
x_train_nb_const = sm.add_constant(x_train_nb, has_constant="add")
x_test_nb_const = sm.add_constant(x_test_nb, has_constant="add")
```

```python
print("y check")
print(y_train_pcp_count.head())
print(y_train_pcp_count.describe())
print(y_train_pcp_count.value_counts().sort_index().head(20))
print("Any NaN in y:", y_train_pcp_count.isna().sum())
print("Any negative y:", (y_train_pcp_count < 0).sum())

print("X check")
print("Any NaN in X:", x_train_nb_const.isna().sum().sum())
print("Any inf in X:", np.isinf(x_train_nb_const.values).sum())
print("X shape:", x_train_nb_const.shape)
```

```python
constant_cols = [
    col for col in x_train_nb_const.columns
    if x_train_nb_const[col].nunique(dropna=False) <= 1
]

print("Constant columns:", constant_cols)

x_train_nb_const2 = x_train_nb_const.drop(columns=constant_cols)
x_test_nb_const2 = x_test_nb_const.drop(columns=[c for c in constant_cols if c in x_test_nb_const.columns])

print(x_train_nb_const2.shape)
```

```python
drop_cols = [
    "Member_Marital_Status_Current_Unknown",
    "Product_Metal_Level_Not Applicable",
    "Product_Type_Not Applicable",
    "Race_Missing"
]

drop_cols = [c for c in drop_cols if c in x_train_nb_const2.columns]

x_train_nb_const2 = x_train_nb_const2.drop(columns=drop_cols)
x_test_nb_const2 = x_test_nb_const2.drop(columns=drop_cols)

print(x_train_nb_const2.shape)
```

```python
poisson_model = sm.GLM(
    y_train_pcp_count,
    x_train_nb_const2,
    family=sm.families.Poisson()
)

poisson_result = poisson_model.fit()

print(poisson_result.summary())
```

```python
y_pred_poisson = poisson_result.predict(x_test_nb_const2)

print("Prediction min/max:", y_pred_poisson.min(), y_pred_poisson.max())

poisson_mae = mean_absolute_error(y_test_pcp_count, y_pred_poisson)
poisson_rmse = np.sqrt(mean_squared_error(y_test_pcp_count, y_pred_poisson))

print("Poisson MAE:", poisson_mae)
print("Poisson RMSE:", poisson_rmse)
```

```python
reference_dummy_cols = [
    "Member_Marital_Status_Current_Unknown",
    "Product_Metal_Level_Not Applicable",
    "Product_Type_Not Applicable",
    "Race_Missing"
]

def prepare_x_for_count_model(x_df, features):
    X = x_df[features].copy()
    X = X.astype(float)

    X = X.replace([np.inf, -np.inf], np.nan)
    X = X.fillna(0)

    X = sm.add_constant(X, has_constant="add")

    constant_cols = [
        col for col in X.columns
        if col != "const" and X[col].nunique(dropna=False) <= 1
    ]

    if len(constant_cols) > 0:
        X = X.drop(columns=constant_cols)

    drop_cols = [c for c in reference_dummy_cols if c in X.columns]

    if len(drop_cols) > 0:
        X = X.drop(columns=drop_cols)

    return X
```

```python
os.makedirs("../results/poisson_intensity_sample", exist_ok=True)

count_outcomes = {
    "pcp_count": "pcp_intensity",
    "er_count": "er_intensity"
}

poisson_summary_rows = []

for outcome_name, outcome_col in count_outcomes.items():
    
    print("====================================================")
    print("Outcome:", outcome_name, "| Column:", outcome_col)
    
    y_train_count = y_train_nb_sample[outcome_col].astype(int)
    y_test_count = y_test_nb_sample[outcome_col].astype(int)

    print("Train distribution:")
    print(y_train_count.value_counts().sort_index().head(20))
    print("Mean:", y_train_count.mean())
    print("Variance:", y_train_count.var())
    print("Variance / Mean:", y_train_count.var() / y_train_count.mean())
    
    for suffix, features in feature_sets.items():
        
        print("----------------------------------------------------")
        print(f"Training Poisson GLM: {outcome_name} | {suffix}")
        print("Number of features:", len(features))
        
        X_train = prepare_x_for_count_model(x_train_nb_sample, features)
        X_test = prepare_x_for_count_model(x_test_nb_sample, features)
        
        # Make sure test has exactly same columns as train
        X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
        
        print("X_train shape:", X_train.shape)
        print("X_test shape:", X_test.shape)
        
        poisson_model = sm.GLM(
            y_train_count,
            X_train,
            family=sm.families.Poisson()
        )
        
        poisson_result = poisson_model.fit()
        
        y_pred_count = poisson_result.predict(X_test)
        
        mae = mean_absolute_error(y_test_count, y_pred_count)
        rmse = np.sqrt(mean_squared_error(y_test_count, y_pred_count))
        
        pred_min = y_pred_count.min()
        pred_max = y_pred_count.max()
        
        # Baseline: always predict train mean
        mean_pred = np.repeat(y_train_count.mean(), len(y_test_count))
        baseline_mae = mean_absolute_error(y_test_count, mean_pred)
        baseline_rmse = np.sqrt(mean_squared_error(y_test_count, mean_pred))
        
        model_name = f"poisson_{outcome_name}_{suffix}_sample"
        
        model_path = f"../results/poisson_intensity_sample/{model_name}.pkl"
        coef_path = f"../results/poisson_intensity_sample/{model_name}_coef_irr.csv"
        summary_path = f"../results/poisson_intensity_sample/{model_name}_summary.txt"
        
        poisson_result.save(model_path)
        
        coef_df = pd.DataFrame({
            "feature": poisson_result.params.index,
            "coef": poisson_result.params.values,
            "IRR": np.exp(poisson_result.params.values),
            "pvalue": poisson_result.pvalues.values
        })
        
        coef_df.to_csv(coef_path, index=False)
        
        with open(summary_path, "w") as f:
            f.write(str(poisson_result.summary()))
        
        poisson_summary_rows.append({
            "outcome": outcome_name,
            "model": suffix,
            "AIC": poisson_result.aic,
            "MAE": mae,
            "RMSE": rmse,
            "baseline_MAE": baseline_mae,
            "baseline_RMSE": baseline_rmse,
            "pred_min": pred_min,
            "pred_max": pred_max,
            "n_train": len(y_train_count),
            "n_test": len(y_test_count),
            "num_features_original": len(features),
            "num_features_used": X_train.shape[1],
            "model_path": model_path,
            "coef_path": coef_path,
            "summary_path": summary_path
        })
        
        print("AIC:", poisson_result.aic)
        print("MAE:", mae)
        print("RMSE:", rmse)
        print("Baseline MAE:", baseline_mae)
        print("Baseline RMSE:", baseline_rmse)
        print("Prediction min/max:", pred_min, pred_max)
        print("Saved model:", model_path)
        print("Saved coefficients:", coef_path)
        
        del X_train, X_test, poisson_model, poisson_result, y_pred_count, coef_df
        gc.collect()
```

```python
poisson_results_df = pd.DataFrame(poisson_summary_rows)

poisson_results_df_round = poisson_results_df.copy()

num_cols = poisson_results_df_round.select_dtypes(include="number").columns
poisson_results_df_round[num_cols] = poisson_results_df_round[num_cols].round(4)

poisson_results_df_round
```
